```python
class MobileNetV3Backbone(nn.Module):
    def __init__(self, backbone, out_channels=256):
        super(MobileNetV3Backbone, self).__init__()
        self.backbone = backbone
        self.out_channels = out_channels
        self.conv = nn.Conv2d(960, self.out_channels, kernel_size=3, stride=1, padding=1)

    def forward(self, x):
        x = self.backbone(x)
        x = self.conv(x)
        return {"0": x}

custom_backbone = MobileNetV3Backbone(backbone=mobilenet_v3.features)
```

```python
class CustomDataset(torch.utils.data.Dataset):
    def __init__(self, json_file, image_dir, category_map=None):
        with open(json_file) as f:
            self.data = json.load(f)
        self.image_dir = image_dir

        self.indices = [img['id'] for img in self.data['images']]
        self.images = {img['id']: img['file_name'] for img in self.data['images']}

        self.annotations = {ann['image_id']: [] for ann in self.data['annotations']}

        if category_map is None:
            category_map = {
                1: 1,
                2: 2,
                5: 5,
                12: 3,
                14: 4,
                16: 6,
                19: 7
            }

        self.category_map = category_map

        for ann in self.data['annotations']:
            x, y, w, h = ann['bbox']
            x2, y2 = x + w, y + h
            mapped_category_id = self.category_map.get(ann['category_id'], ann['category_id'])
            self.annotations[ann['image_id']].append({
                'bbox': [x, y, x2, y2],
                'category_id': mapped_category_id
            })

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, indx):
        image_id = self.indices[indx]
        image_path = f"{self.image_dir}/{self.images[image_id]}"

        img = Image.open(image_path)

        mean = [0.5]
        std = [0.5]

        transform = T.Compose([
            T.Grayscale(num_output_channels=3),
            T.ToTensor(),
            T.Normalize(mean=mean * 3, std=std * 3)
            ])


        img = transform(img)

        annotations = self.annotations[image_id]

        boxes = [ann['bbox'] for ann in annotations]
        boxes_tensor = torch.tensor(boxes, dtype=torch.float32)

        labels = torch.tensor([ann['category_id'] for ann in annotations], dtype=torch.int64)

        target = {'boxes': boxes_tensor, 'labels': labels}

        return img, target

def coco_collate_fn(batch):
    return batch
```

```python
def compute_class_weights(dataset, num_classes):
    class_counts = np.zeros(num_classes)
    total_samples = 0

    for data in dataset:
        labels = data['labels']
        class_counts[labels] += 1
        total_samples += len(labels)

    class_frequencies = class_counts / total_samples
    class_weights = 1.0 / (class_frequencies + 1e-8)
    class_weights = class_weights / np.sum(class_weights) * num_classes
    return torch.tensor(class_weights, dtype=torch.float32)

train_labels = []
for batch in train_loader:
    images, targets = zip(*batch)
    for target in targets:
        train_labels.append(target)

num_classes = 8
class_weights = compute_class_weights(train_labels, num_classes)
```

```python
class_weights = class_weights.to(device)
```

```python
class WeightedCrossEntropyLoss(nn.Module):
    def __init__(self, class_weights=None, reduction='mean'):
        super(WeightedCrossEntropyLoss, self).__init__()
        self.class_weights = class_weights
        self.reduction = reduction

    def forward(self, logits, targets):
        """
        logits: [batch_size, num_boxes, num_classes]
        targets: [batch_size, num_boxes] (Class indices for each bounding box)
        """
        batch_size = logits.size(0)
        num_boxes = logits.size(1)
        num_classes = logits.size(2)

        log_probs = F.log_softmax(logits, dim=-1)

        if self.class_weights is not None:
            weights = self.class_weights[None, None, :]
            loss = -weights * torch.gather(log_probs, dim=-1, index=targets.unsqueeze(-1))
        else:
            loss = -torch.gather(log_probs, dim=-1, index=targets.unsqueeze(-1))

        if self.reduction == 'mean':
            return loss.mean()
        elif self.reduction == 'sum':
            return loss.sum()
        else:
            return loss
```

```python
loss_fn = WeightedCrossEntropyLoss(class_weights=class_weights, reduction='mean')
```

```python
for epoch in range(epoches):
    model.train()
    running_loss = 0.0
    print(f"Epoch {epoch + 1}/{epoches}")
    progress_bar = tqdm(train_loader, desc=f"Training", ncols=80)

    for batch in progress_bar:
        images, targets = zip(*batch)
        images = [image.to(device) for image in images]
        targets = [{k: v.to(device) for k, v in t.items()} for t in targets]

        optimizer.zero_grad()
        loss_dict = model(images, targets)

        classification_loss = loss_dict['loss_classifier']
        regression_loss = loss_dict['loss_box_reg']

        focal_loss = classification_loss * class_weights
        total_loss = focal_loss + regression_loss

        total_loss = total_loss.sum()

        total_loss.backward()
        optimizer.step()

        running_loss += total_loss.item()
        progress_bar.set_postfix({
            "loss": f"{total_loss.item():.4f}"
        })

    avg_train_loss = running_loss / len(train_loader)
    print(f"Epoch {epoch + 1}/{epoches} - Training Loss: {avg_train_loss:.4f}")

    scheduler.step()

    model.eval()
    with torch.no_grad():
        running_iou_loss = 0.0
        total_batches = 0

        for batch in val_loader:
            try:
                images, targets = zip(*batch)
                images = [img.to(device) for img in images]
                targets = [{k: v.to(device) for k, v in t.items()} for t in targets]

                outputs = model(images)
                iou_loss = 0.0
                for o, t in zip(outputs, targets):
                    if o["boxes"].size(0) == 0 or t["boxes"].size(0) == 0:
                        continue
                    min_size = min(o["boxes"].size(0), t["boxes"].size(0))
                    iou_loss += F.smooth_l1_loss(o["boxes"][:min_size], t["boxes"][:min_size])

                if len(outputs) > 0:
                    iou_loss /= len(outputs)
                    running_iou_loss += iou_loss.item()
                    total_batches += 1

            except RuntimeError as e:
                print(f"Error during validation on batch: {e}")
                continue

        if total_batches > 0:
            avg_iou_loss = running_iou_loss / total_batches
        else:
            avg_iou_loss = 0.0
        print(f"Epoch {epoch + 1}/{epoches} - Validation IoU Loss: {avg_iou_loss:.4f}")

    model_save_path = os.path.join(model_save_folder, f'model_epoch_{epoch + 1}.pth')
    torch.save({
        'epoch': epoch + 1,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'loss': avg_train_loss,
    }, model_save_path)

print(f"Training Completed")
```

```python
import torchvision.ops as ops

confidence_threshold = 0.3

all_preds = []
all_targets = []

with torch.no_grad():
    for batch in tqdm(test_loader, desc="Evaluating"):
        images, targets = zip(*batch)
        images = [img.to(device) for img in images]

        outputs = model(images)

        for output, target in zip(outputs, targets):
            high_conf_ts = output['scores'] >= confidence_threshold

            filtered_output = {
                'boxes' : output['boxes'][high_conf_ts].cpu(),
                'labels' : output['labels'][high_conf_ts].cpu(),
                'scores' : output['scores'][high_conf_ts].cpu()
            }
            if filtered_output['boxes'].size(0) > 0:
                all_preds.append(filtered_output)
            all_targets.append({
                'boxes': target['boxes'].cpu(),
                'labels': target['labels'].cpu()
            })
```
